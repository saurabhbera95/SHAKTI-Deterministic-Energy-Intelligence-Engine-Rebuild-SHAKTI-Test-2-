# SHAKTI Deterministic Energy Intelligence Engine  
## (TANTRA – Test 2)

This notebook implements a fully executable, deterministic energy intelligence pipeline.
No machine learning is used. All outputs are rule-based and traceable. 

Executable, traceable, deterministic pipeline (NO ML)

```python
# Import required libraries
# json is used for structured output formatting

import json
```

```python
# Simulating multi-source data input
# In a real system, this would come from sensors, logs, APIs, etc.

sensor_data = {
    "device_id": "S1",
    "temperature": 60,        # High temperature (should trigger rule)
    "energy_output": 18       # Low output (should trigger rule)
}

logs_data = "[ALERT] Efficiency dropped significantly"  # Log-based signal
```

```python
# Ingestion layer: collects raw data from different sources

def ingest_data():
    # Returns sensor and log data
    return sensor_data, logs_data

# Execute ingestion
sensor, logs = ingest_data()

# Display ingested data
sensor, logs
```

```python
# Classification layer: tags data based on source and type

def classify_sensor(data):
    return {
        "source": "sensor",
        "type": "energy",
        "device_id": data["device_id"]
    }

def classify_logs(logs):
    return {
        "source": "logs",
        "type": "text"
    }

# Apply classification
sensor_meta = classify_sensor(sensor)
logs_meta = classify_logs(logs)

sensor_meta, logs_meta
 
```

```python
# Cleaning layer: ensures consistent data format

def clean_sensor(data):
    # Convert values to float for standardization
    data["temperature"] = float(data["temperature"])
    data["energy_output"] = float(data["energy_output"])
    return data

# Apply cleaning
sensor = clean_sensor(sensor)

sensor
```

```python
# Parsing layer: extracts structured information from unstructured logs

def parse_logs(log_text):
    log_text = log_text.lower()

    # Detect specific events from logs
    if "efficiency dropped" in log_text:
        return {"event": "EFFICIENCY_DROP"}

    return {}

# Apply parsing
parsed_logs = parse_logs(logs)

parsed_logs
 
```

```python
# Core logic: deterministic rule engine (NO ML)
# Generates signals based on predefined thresholds

def apply_rules(sensor, parsed_logs):
    signals = []
    trace = []   # Stores which rule triggered (for auditability)

    # Rule 1: High temperature
    if sensor["temperature"] > 50:
        signals.append("OVERHEAT")
        trace.append("Rule Triggered: temperature > 50")

    # Rule 2: Low energy output
    if sensor["energy_output"] < 25:
        signals.append("LOW_OUTPUT")
        trace.append("Rule Triggered: energy_output < 25")

    # Rule 3: Log-based detection
    if parsed_logs.get("event") == "EFFICIENCY_DROP":
        signals.append("EFFICIENCY_DROP")
        trace.append("Rule Triggered: log detected efficiency drop")

    return signals, trace

# Apply rules
signals, trace = apply_rules(sensor, parsed_logs)

signals, trace
 
```

```python
# Routing layer: decides which team/action is triggered based on signals

def route(signals):
    actions = []

    if "OVERHEAT" in signals:
        actions.append("SEND_TO_MAINTENANCE")

    if "LOW_OUTPUT" in signals:
        actions.append("NOTIFY_OPERATIONS")

    if "EFFICIENCY_DROP" in signals:
        actions.append("CHECK_SYSTEM")

    return actions

# Apply routing
actions = route(signals)

actions
 
```

```python
# Output layer: generates structured JSON output
# This is critical for system verification and integration

def generate_output(device_id, signals, actions, trace):
    return {
        "device_id": device_id,
        "signals": signals,
        "actions": actions,
        "trace": trace   # Ensures full explainability
    }

# Generate final output
output = generate_output(sensor["device_id"], signals, actions, trace)

output 
 
```

```python
# Display output in readable JSON format

print(json.dumps(output, indent=4))
 
```

## Traceability

- Every signal is generated using deterministic rules  
- Each rule is recorded in the trace  
- No black-box logic is used  
- Output is fully explainable and auditable
